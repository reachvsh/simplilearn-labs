## 2. Save the course API key (one time)

Paste the key your instructor shared in class. It is saved into the `.env` file in this folder (hidden in the file browser because the name starts with a dot). `sync.py` never overwrites it, and every lab reads it from there.

```python
import os, re, getpass
from pathlib import Path

env_file = Path(".env")
text = env_file.read_text() if env_file.exists() else ""
if re.search(r"^OPENAI_API_KEY=sk-", text, re.M):
    print("Key already saved in .env. Nothing to do.")
elif os.getenv("OPENAI_API_KEY"):
    print("An API key is already set in this environment. Nothing to do.")
else:
    key = getpass.getpass("Paste the course API key: ").strip()
    env_file.write_text(text.rstrip("\n") + f"\nOPENAI_API_KEY={key}\n")
    print("Saved to .env")
```
